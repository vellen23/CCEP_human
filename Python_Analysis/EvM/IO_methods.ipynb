```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import trapz
```

```python
plt.rcParams.update({
            'font.family': 'arial',
            'font.size': 12,
            'xtick.labelsize': 8,
            'ytick.labelsize': 8,
            'legend.fontsize': 9,
            'svg.fonttype': 'none',
            'font.size': 10,
            'axes.titlesize': 10,
            'axes.labelsize': 8,
            'xtick.labelsize': 8,
            'ytick.labelsize': 8,
            'legend.fontsize': 9,
            'figure.titlesize': 10
        })
```

```python

# Define x values (normalized)
x = np.linspace(0, 1, 100)

# Define linear y values
y_linear = x

# Define sigmoidal y values with different AUCs
y_sigmoid_1 = 1 / (1 + np.exp(-10 * (x - 0.5)))  # AUC = 0.5
y_sigmoid_2 = 1 / (1 + np.exp(-10 * (x - 0.25)))  # AUC = 0.75
y_sigmoid_3 = 1 / (1 + np.exp(-10 * (x - 0.6)))  # AUC = 0.25
y_sigmoid_2[0]= 0
y_sigmoid_2[-1]= 1
y_sigmoid_3[0]= 0
y_sigmoid_3[-1]= 1
# Calculate AUC for each sigmoidal curve
auc_1 = trapz(y_sigmoid_1, x)
auc_2 = trapz(y_sigmoid_2, x)
auc_3 = trapz(y_sigmoid_3, x)

# Create subplots
fig, axs = plt.subplots(1, 4, figsize=(8, 3), sharey=True)
plt.suptitle('Possible GAIN-Curves')
# Plot linear relationship
axs[0].plot(x, y_linear, color='k')
axs[0].fill_between(x, 0, y_linear, color='gray', alpha=0.5)
axs[0].set_title(f'Linear (AUC = 0.5)')
axs[0].set_ylabel(f'Normalized Response Strength')


# Plot sigmoidal relationships
axs[1].plot(x, y_sigmoid_1, color='k')
axs[1].fill_between(x, 0, y_sigmoid_1, color='gray', alpha=0.5)
axs[1].set_title(f'Sigmoidal (AUC = {auc_1:.2f})')

axs[2].plot(x, y_sigmoid_2, color='k')
axs[2].fill_between(x, 0, y_sigmoid_2, color='gray', alpha=0.5)
axs[2].set_title(f'Sigmoidal (AUC = {auc_2:.2f})')

axs[3].plot(x, y_sigmoid_3, color='k')
axs[3].fill_between(x, 0, y_sigmoid_3, color='gray', alpha=0.5)
axs[3].set_title(f'Sigmoidal (AUC = {auc_3:.2f})')

for i in range(4):
    axs[i].set_xlabel(f'Normalized Stimulation Intensity')
    axs[i].set_aspect(1)
# Adjust layout
plt.tight_layout()
filename = 'X:\\4 e-Lab\EvM\Projects\EL_experiment\Analysis\Supp_figures\IO\General\AUC_Theory.svg'
plt.savefig(filename)
```
