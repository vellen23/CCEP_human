```python
def group_connections(df):
    """
    Group connections into 'local', 'direct', or 'indirect' based on certain conditions.

    - 'local': Connections with d <= 20, h == 0, and not part of the forbidden pairs.
    - 'direct': Non-local connections with delay <= 20.
    - 'indirect': Non-local connections with delay > 20.

    Parameters:
    df (pd.DataFrame): DataFrame containing the connections with columns 'd', 'StimR', 'ChanR', 'H', and 'delay'.

    Returns:
    pd.DataFrame: DataFrame with a new column 'Group' indicating the group of each connection.
    """

    # Define impossible local pairs
    impossible_local_pairs = set([('Frontal', 'Temporal'), ('Frontal', 'Basotemporal')])

    # Initialize all connections as 'direct' first
    df['Group'] = 'direct'

    # Conditions for local connections
    is_local = (df['d'] <= 20) & (df['H'] == 0)
    for pair in impossible_local_pairs:
        is_local &= (~((df['StimR'] == pair[0]) & (df['ChanR'] == pair[1]))) & (~((df['ChanR'] == pair[0]) & (df['StimR'] == pair[1])))

    # Assign 'local' to the filtered connections
    df.loc[is_local, 'Group'] = 'local'

    # Assign 'indirect' to non-local connections with delay > 20
    df.loc[(df['Group'] == 'direct') & (df['delay'] > 20), 'Group'] = 'indirect'

    return df


def adding_anatomy(df, pair=1, area='Destrieux'):
    # area == 'Region' or 'Area'
    df = df.reset_index(drop=True)
    CIRC_AREAS_FILEPATH = 'X:\\4 e-Lab\e-Lab shared code\Softwares\Connectogram\circ_areas.xlsx'
    atlas = pd.read_excel(CIRC_AREAS_FILEPATH, sheet_name='atlas')
    if pair:
        for subregion in np.unique(df[['StimA', 'ChanA']]):
            region = atlas.loc[atlas.Abbreviation == subregion, area].values

            if len(region) > 0:
                if region[0] == "entorhinal_cortex":
                    region[0] = 'Entorhinal'
                elif region[0] == "AnteroHippocampus":
                    region[0] = 'Hippocampus'
                elif region[0] == "PosteroHippocampus":
                    region[0] = 'Hippocampus'
                df.loc[df.StimA == subregion, 'StimR'] = region[0]
                df.loc[df.ChanA == subregion, 'ChanR'] = region[0]
            else:
                region = atlas.loc[atlas.Subregion == subregion, area].values
                if len(region) > 0:
                    if region[0] == "entorhinal_cortex":
                        region[0] = 'Entorhinal'
                    elif region[0] == "AnteroHippocampus":
                        region[0] = 'Hippocampus'
                    elif region[0] == "PosteroHippocampus":
                        region[0] = 'Hippocampus'
                    df.loc[df.StimA == subregion, 'StimR'] = region[0]
                    df.loc[df.ChanA == subregion, 'ChanR'] = region[0]
                else:
                    df.loc[df.StimA == subregion, 'StimR'] = 'U'
                    df.loc[df.ChanA == subregion, 'ChanR'] = 'U'
    else:
        for subregion in np.unique(df[['ChanA']]):
            region = atlas.loc[atlas.Abbreviation == subregion, area].values
            if len(region) > 0:
                df.loc[df.ChanA == subregion, 'ChanR'] = region[0]
            else:
                df.loc[df.ChanA == subregion, 'ChanR'] = 'U'
    return df
```

```python
def adding_distance_tracts(df, df_atlas, atlas=False):
    # Correct renaming of columns with a dictionary and using inplace or assignment
    df_atlas = df_atlas.rename(columns={'Count': 'tract_num', 'Length': 'tract_dist'})
    for col in ['tract_num', 'tract_dist']:
        if col in df:
            df = df.drop(columns=col)
    # Ensure that 'Stim' and 'Chan' are columns to be merged on both DataFrames
    # Correct the usage of selecting multiple columns from df_atlas
    if atlas:
        df = df.merge(df_atlas, on=['Stim', 'Chan'],
                      how='left').reset_index(drop=True)

    else:
        df = df.merge(df_atlas[['Stim', 'Chan', 'tract_num', 'tract_dist']], on=['Stim', 'Chan'],
                      how='left').reset_index(drop=True)
    return df
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.optimize import curve_fit
from scipy.stats import ttest_ind

def logistic_model(x, K, P0, r):
    """Logistic growth model."""
    return K / (1 + ((K - P0) / P0) * np.exp(-r * x))

def fit_logistic(data, x_col, y_col):
    """Fit the logistic model and return parameters."""
    popt, _ = curve_fit(logistic_model, data[x_col], data[y_col], p0=[1, 0.01, 0.02])
    return popt

def bootstrap_params_logistic(data, x_col, y_col, n=1000):
    """Bootstrap the parameter estimation for logistic."""
    boot_params = []
    for _ in range(n):
        sample = data.sample(frac=1, replace=True)
        popt = fit_logistic(sample, x_col, y_col)
        boot_params.append(popt)
    return np.array(boot_params)
# Fit models for both groups
group_s_minus = data_e[data_e['WM_tract'] == 's-']
group_s_plus = data_e[data_e['WM_tract'] == 's+']

params_s_minus = fit_logistic(group_s_minus, 'd', 'Sig')
params_s_plus = fit_logistic(group_s_plus, 'd', 'Sig')

# Generate bootstrap samples
boot_params_s_minus = bootstrap_params_logistic(group_s_minus, 'd', 'Sig')
boot_params_s_plus = bootstrap_params_logistic(group_s_plus, 'd', 'Sig')

# Plotting for visual comparison
x = np.linspace(min(data_e['d']), max(data_e['d']), 100)
y_s_minus = logistic_model(x, *params_s_minus)
y_s_plus = logistic_model(x, *params_s_plus)


# T-tests on the parameters
t_test_K = ttest_ind(boot_params_s_minus[:, 0], boot_params_s_plus[:, 0])
t_test_P0 = ttest_ind(boot_params_s_minus[:, 1], boot_params_s_plus[:, 1])
t_test_r = ttest_ind(boot_params_s_minus[:, 2], boot_params_s_plus[:, 2])
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import ttest_ind

def fit_exponential(data, x_col, y_col):
    """Fit the exponential model and return parameters."""
    fit = np.polyfit(data[x_col], np.log(data[y_col]), 1)
    a = np.exp(fit[1])  # intercept
    b = fit[0]          # slope
    return a, b

def bootstrap_params(data, x_col, y_col, n=1000):
    """Bootstrap the parameter estimation."""
    params = np.array([fit_exponential(data.sample(frac=1, replace=True), x_col, y_col) for _ in range(n)])
    return params

# Fit models for both groups
group_s_minus = data_e[data_e['WM_tract'] == 's-']
group_s_plus = data_e[data_e['WM_tract'] == 's+']

a_s_minus, b_s_minus = fit_exponential(group_s_minus, 'd', 'Sig')
a_s_plus, b_s_plus = fit_exponential(group_s_plus, 'd', 'Sig')

# Generate bootstrap samples
params_s_minus = bootstrap_params(group_s_minus, 'd', 'Sig')
params_s_plus = bootstrap_params(group_s_plus, 'd', 'Sig')

# Plotting for visual comparison
x = np.linspace(0, 100, 100)
y_s_minus = a_s_minus * np.exp(b_s_minus * x)
y_s_plus = a_s_plus * np.exp(b_s_plus * x)


```

```python
def ax_violon_box_strip(ax, x, y, data):
    palette= sns.light_palette("seagreen", as_cmap=True)
    sns.violinplot(x=x, y=y, data=data, dodge=False,
                    scale="width", inner=None, ax = ax, palette='Blues')
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    for violin in ax.collections:
        bbox = violin.get_paths()[0].get_extents()
        x0, y0, width, height = bbox.bounds
        violin.set_clip_path(plt.Rectangle((x0, y0), width / 2, height, transform=ax.transData))

    sns.boxplot(x=x, y=y, data=data, saturation=1, showfliers=False,
                width=0.3, boxprops={'zorder': 3, 'facecolor': 'none'}, ax=ax)
    old_len_collections = len(ax.collections)
    sns.stripplot(x=x, y=y, data=data, dodge=False, ax=ax, color='k', alpha = 0.1, s=1)
    for dots in ax.collections[old_len_collections:]:
        dots.set_offsets(dots.get_offsets() + np.array([0.12, 0]))
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
    return ax
```

```python
# Create figure and gridspec layout
fig = plt.figure(figsize=(8, 3))
gs_master = gridspec.GridSpec(1, 3, width_ratios=[1, 1, 1])  # Master grid

# Normal subplots on the sides
ax0 = fig.add_subplot(gs_master[0])
ax2 = fig.add_subplot(gs_master[2])

# Split middle plot with a broken y-axis
gs_middle = gridspec.GridSpecFromSubplotSpec(2, 1, subplot_spec=gs_master[1], 
                                             height_ratios=[1, 3],
                                             hspace=0.03)

ylim = [400, 12000]
ylim2 = [0, 400]

ax1 = fig.add_subplot(gs_middle[0])
ax1_2 = fig.add_subplot(gs_middle[1])

# Plot data in all subplots
for i, (x, xlab) in enumerate(zip(x_sel, x_labels)):
    if i == 1:  # Middle subplot
        sns.histplot(x=x, data=df_WM, hue='WM_tract', multiple='stack', ax=ax1, legend=False, bins = 50)
        sns.histplot(x=x, data=df_WM, hue='WM_tract', multiple='stack', ax=ax1_2, legend=False, bins = 50)
        ax1.set_ylim(ylim)
        ax1_2.set_ylim(ylim2)
        ax1.spines['bottom'].set_visible(False)
        ax1_2.spines['top'].set_visible(False)
        ax1.xaxis.tick_top()
        ax1.tick_params(labeltop=False)  # hide the tick labels on top
        ax1_2.xaxis.tick_bottom()
    else:
        ax = ax0 if i == 0 else ax2
        sns.histplot(x=x, data=df_WM, hue='WM_tract', multiple='stack', legend=False, ax=ax, bins = 50)
        ax.set_xlabel(x_labels[i])
        ax.set_ylabel('Count')
ax.legend(['Contact','Destrieux','None'], title='WM tract')
plt.tight_layout()  # Adjust layout
plt.show()
```
