```python
# Number of blocks
n_blocks = int(len(np.unique(df['Block'])))

# Create empty list to hold matrices
block_matrices = []

# Loop through each block and create a matrix
for block in np.unique(df['Block']):
    # Filter the DataFrame for the current block
    block_data = df[df['Block'] == block]
    
    # Create an empty DataFrame with Stim as rows, Chan as columns, filled with NaN
    matrix = pd.DataFrame(np.nan, index=range(len(labels_all)), columns=range(len(labels_all)))
    
    # Populate the matrix with mean LL values
    for index, data in block_data.groupby(['Stim', 'Chan']):
        matrix.at[index[0], index[1]] = data['LL_sig'].mean()
    
    # Append the matrix to the list
    block_matrices.append(matrix)

```

```python
# Initialize an empty correlation matrix
n = len(block_matrices)
corr_matrix = np.zeros((n, n))

# Compute the correlation matrix, handling NaNs
for i in range(n):
    for j in range(i+1, n):
        # Mask NaN values
        mask_i = np.isnan(block_matrices[i])
        mask_j = np.isnan(block_matrices[j])
        combined_mask = mask_i | mask_j

        # Create masked arrays
        x = np.ma.array(block_matrices[i].values.flatten(), mask=combined_mask.values.flatten())
        y = np.ma.array(block_matrices[j].values.flatten(), mask=combined_mask.values.flatten())

        # Compute correlation while ignoring masked (NaN) entries
        corr = np.ma.corrcoef(x, y)

        # Fill the correlation matrix with the computed values
        corr_matrix[i, j] = corr[0, 1]
        corr_matrix[j, i] = corr[0, 1]  # Correlation matrix is symmetric

# Handle the diagonal (correlation with itself is always 1)
np.fill_diagonal(corr_matrix, 1)
```

```python
def symnmf_newton(A, k, maxiter=10000, tol=1e-4, sigma=0.1, beta=0.1, computeobj=True, debug=0):
    n = A.shape[0]
    if A.shape[0] != A.shape[1]:
        raise ValueError('A must be a symmetric matrix!')

    # Initialize H
    H = 2 * np.sqrt(np.mean(A) / k) * np.random.rand(n, k)

    # Initialize objective function
    obj = np.linalg.norm(A - H.dot(H.T), 'fro') ** 2
    gradH = 4 * (H.dot(H.T).dot(H) - A.dot(H))
    initgrad = np.linalg.norm(gradH, 'fro')
    
    if debug:
        print('init grad norm %g' % initgrad)

    for iter in range(1, maxiter + 1):
        gradH = 4 * (H.dot(H.T).dot(H) - A.dot(H))
        projnorm_idx = np.logical_or(gradH <= 0, H > 0)
        projnorm = np.linalg.norm(gradH[projnorm_idx])
        
        if projnorm < tol * initgrad:
            if debug:
                print('final grad norm %g' % projnorm)
            break
        elif debug > 1:
            print('iter %d: grad norm %g' % (iter, projnorm))

        # Compute Newton step
        step = np.zeros((n, k))
        for i in range(k):
            if np.any(projnorm_idx[:, i]):
                H_subset = H[projnorm_idx[:, i], :]
                hessian = 4 * ((H_subset.dot(H_subset.T)) + np.outer(H[:, i], H[:, i]) + np.eye(projnorm_idx[:, i].sum()))
                try:
                    step[projnorm_idx[:, i], i] = np.linalg.solve(hessian, -gradH[projnorm_idx[:, i], i])
                except np.linalg.LinAlgError:
                    step[:, i] = -gradH[:, i]
            else:
                step[:, i] = -gradH[:, i]

        # Line search
        alpha_newton = 1
        while True:
            H_new = np.maximum(H - alpha_newton * step, 0)
            new_obj = np.linalg.norm(A - H_new.dot(H_new.T), 'fro') ** 2
            if new_obj - obj <= sigma * np.sum(gradH * (H_new - H)):
                H = H_new
                obj = new_obj
                break
            alpha_newton *= beta

    return H, iter, obj

# Example usage:
# A is your symmetric matrix, k is the number of clusters
# H, iter, obj = symnmf_newton(A, k)
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import NMF
from sklearn.preprocessing import MinMaxScaler

k = 2

# Ensure corr_matrix is a numpy array, handling NaN by filling them with 0
# Symmetric NMF doesn't handle NaN, so we replace NaNs with the mean of the matrix
nan_mask = np.isnan(corr_matrix)
mean_val = np.nanmean(corr_matrix)
corr_matrix_filled = np.where(nan_mask, mean_val, corr_matrix)

# Scale the filled correlation matrix to be non-negative since NMF requires non-negative inputs
scaler = MinMaxScaler()
corr_matrix_scaled = scaler.fit_transform(corr_matrix_filled)

# Perform NMF
model = NMF(n_components=k, init='random', random_state=0)
W = model.fit_transform(corr_matrix_scaled)
H = model.components_

# Plot the original correlation matrix and the matrix H
fig, axes = plt.subplots(1, 3, figsize=(12, 6), gridspec_kw={'width_ratios': [3, 0.45, 3]})

# Plot the correlation matrix
sns.heatmap(corr_matrix_filled, ax=axes[0], cmap='viridis', vmin =0.8, vmax = 0.95)
axes[0].set_title('Original Correlation Matrix')
axes[0].set_xlabel('Blocks')
axes[0].set_ylabel('Blocks')
axes[0].set_aspect('equal', 'box')
# Plot the matrix H
sns.heatmap(H, ax=axes[2], cmap='hot', vmin  =0.2)
axes[2].set_title('NMF Components (H)')
axes[2].set_xlabel('Clusters')
axes[2].set_ylabel('Blocks')
axes[2].set_aspect(5, 'box')
# Plot the matrix H
sns.heatmap(W, ax=axes[1], cmap='hot', vmin  =0.2)
axes[1].set_title('NMF Components (W)')
axes[1].set_ylabel('Clusters')
axes[1].set_xlabel('Blocks')
axes[1].set_aspect(1/5, 'box')

plt.tight_layout()
plt.show()
```
